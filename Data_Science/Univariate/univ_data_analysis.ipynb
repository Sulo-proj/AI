```python
import pandas as pd
from data_analysis_utils import Univariate, Preprocessing, NormalDistribution

```

```python

data = pd.read_csv('Placement.csv')
data.drop(columns = 'sl_no', axis = 1, inplace = True)

quan, qual = Univariate.quanQual(data)

print(quan, qual)
```

### Mean, Mode, Median, Percentile, IQR, Skew, Kurtosis, Var, Std


```python
descriptive = Univariate.descriptive(data, quan)

print(descriptive)
```

```python
descriptive, lesser, greater = Univariate.detect_outliers(descriptive, quan)

print(lesser, greater)
```

```python
data = Univariate.remove_outliers(descriptive, data, lesser, greater)
```

```python
descriptive_after_change = Univariate.descriptive(data, quan)

descriptive_after_change, lesser, greater = Univariate.detect_outliers(descriptive_after_change, quan)

print(lesser, greater)
```

# Preprocessing Data

```python
data = Preprocessing.simple_missing(data, quan, qual)
data = Preprocessing.model_missing(data, quan)
```

```python
startrange = 50
endrange = 70
perc_lookup = 80
columnName = 'ssc_p'
sample_mean, sample_std, appx_prob, actual_prob, ecdf = NormalDistribution.get_pdf_probability(data[columnName], startrange, endrange, perc_lookup)

print('Column: {}'.format(columnName))
print('\nMean = %.3f, Standard Deviation = %.3f' % (sample_mean, sample_std))
print("\nThe area between range ({},{}): Approximately : {:.3f}, Actually: {:.3f}".format(startrange, endrange, appx_prob, actual_prob))
print("\nECDF (Percentile Look Up) at {} : {:.3f}%".format(perc_lookup, ecdf(perc_lookup)*100))


```

```python
avg_z, std_z = NormalDistribution.stdNBgraph(data[columnName], sample_mean, sample_std)

print("\nAverage of z-scores:", avg_z.round(3))
print("\nStandard deviation of z-scores:", std_z.round(3))
```
