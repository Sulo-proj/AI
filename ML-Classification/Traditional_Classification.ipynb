```python
#importing the Libraies
import numpy as np
# import matplotlib.pyplot as plt
import pandas as pd
import classification
from sklearn.model_selection import train_test_split

```

```python
dataset=pd.read_csv("Social_Network_Ads.csv")
dataset=pd.get_dummies(dataset,drop_first=True)
dataset=dataset.drop("User ID",axis=1)
dataset["Purchased"].value_counts()
dataset
```

```python
indep=dataset[["Age","EstimatedSalary","Gender_Male"]]
dep=dataset["Purchased"]
```

```python

model_name = 'LogisticRegression'
params = {'solver':'lbfgs', 'penalty':'l2', 'C':1.0, 'max_iter':100, 'random_state':10}
logistic_classifier=classification.classification_traditional(indep,dep, model_name, **params)

model_name = 'KNeighborsClassifier'
params = {'n_neighbors': 7, 'metric': 'minkowski', 'p' : 2}
knn_classifier=classification.classification_traditional(indep,dep, model_name, **params)

model_name = 'RandomForestClassifier'
params = {'n_estimators': 100, 'criterion': 'entropy', 'random_state': 10}
rf_classifier=classification.classification_traditional(indep,dep, model_name, **params)

model_name = 'DecisionTreeClassifier'
params = {'criterion': 'entropy', 'random_state': 10}
dt_classifier=classification.classification_traditional(indep,dep, model_name, **params)

model_name = 'SVC'
params = {'kernel': 'rbf', 'C': 1.0, 'gamma': 'scale', 'random_state': 10}
svc_classifier=classification.classification_traditional(indep,dep, model_name, **params)

model_name = 'MultinomialNB'
multinb_classifier=classification.classification_traditional(indep,dep, model_name )

model_name = 'CategoricalNB'
categnb_classifier=classification.classification_traditional(indep,dep, model_name)

model_name = 'ComplementNB'
complmntnb_classifier=classification.classification_traditional(indep,dep, model_name)

model_name = 'BernoulliNB'
bernoullinb_classifier=classification.classification_traditional(indep,dep, model_name)

model_name = 'GaussianNB'
guassian_classifier=classification.classification_traditional(indep,dep, model_name)

model_name = 'AdaBoostClassifier'
params = {'n_estimators': 50, 'learning_rate': 1.0, 'random_state': 10}
adaboost_classifier=classification.classification_traditional(indep,dep, model_name, **params)

model_name = 'GradientBoostingClassifier'
params = {'n_estimators': 100, 'learning_rate': 0.1, 'max_depth': 3, 'random_state': 10}
gradientboost_classifier=classification.classification_traditional(indep,dep, model_name, **params)

model_name = 'XGBClassifier'
params = {'n_estimators': 100, 'learning_rate': 0.1, 'max_depth': 3, 'random_state': 10}
xgb_classifier=classification.classification_traditional(indep,dep, model_name, **params)

model_name = 'LGBMClassifier'
params = {'n_estimators': 100, 'learning_rate': 0.1, 'verbosity':-1, 'random_state': 10}
lgbm_classifier=classification.classification_traditional(indep,dep, model_name, **params)

```
