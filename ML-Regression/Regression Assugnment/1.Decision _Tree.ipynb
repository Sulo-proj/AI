```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn import tree
from sklearn.metrics import r2_score
```

```python
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
print(dataset.columns)
dataset
```

```python
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
independent
```

```python
dependent=dataset[["charges"]]
```

```python
X_train,X_test,y_train,y_test=train_test_split(independent, dependent, test_size=0.25,random_state=10)
```

```python
regressor=DecisionTreeRegressor(criterion='friedman_mse',max_features='log2', splitter='best', random_state=10)
regressor=regressor.fit(X_train,y_train)
```

```python
tree.plot_tree(regressor)
plt.show()
```

```python
y_pred=regressor.predict(X_test)
r_score=r2_score(y_test,y_pred)
r_score
```
