```python
import pandas as pd
import regression_utils  
```

```python
# Load and prepare dataset
dataset = pd.read_csv("insurance_pre.csv")
dataset = pd.get_dummies(dataset, drop_first=True)
X = dataset.drop("charges", axis=1).values
y = dataset["charges"].values
print("X shape:", X.shape, "y shape:", y.shape)
```

```python

# Dictionary to define models and their param grids
models_with_params = {
    'LinearRegression': None,
    'Ridge': {'alpha': [0.01, 0.1, 1, 10, 100]},
    'Lasso': {'alpha': [0.001, 0.01, 0.1, 1, 10]},
    'ElasticNet': {
        'alpha': [0.001, 0.01, 0.1, 1],
        'l1_ratio': [0.1, 0.5, 0.7, 0.9]
    },
    'QuantileRegressor': {
        'alpha': [ 0.1, 1, 10, 100],
        'quantile': [0.1, 0.5, 0.9],
        'solver': ['highs', 'highs-ds', 'highs-ipm']
        # 'solver': ['highs', 'highs-ds', 'highs-ipm', 'interior-point', 'revised simplex']
    },
    'SVR': {
        'kernel': ['linear', 'poly', 'rbf', 'sigmoid'],
        'C': [50, 100, 500, 1000, 2000, 3000],
        'gamma': ['auto', 'scale']
    },
    'DecisionTreeRegressor': {
        'criterion': ['squared_error', 'absolute_error', 'friedman_mse', 'poisson'],
        'max_features': [None, 'sqrt', 'log2'],
        'splitter': ['best', 'random']
    },
    'RandomForestRegressor': {
        'criterion': ['absolute_error', 'friedman_mse', 'poisson', 'squared_error'],
        'max_features': [None, 'sqrt', 'log2'],
        'n_estimators': [10, 50, 100]
    },
    'GradientBoostingRegressor': {
        'n_estimators': [100, 200],
        'learning_rate': [0.01, 0.1],
        'max_depth': [3, 4, 5]
    },
    'SGDRegressor': {
        'alpha': [0.0001, 0.001, 0.01, 0.1, 1, 10, 100],
        'max_iter': [1000, 2000, 3000],
        'loss': ['squared_error', 'huber', 'epsilon_insensitive', 'squared_epsilon_insensitive']
    }
}


```

```python
# Train models and store scores
all_models_score = {}
for model_name, param_grid in models_with_params.items():
    print('---------------------------------------------------------------------------')
    print(f"\n===== Running model: {model_name} =====")
    model, score, _ = regression_utils.run_regression_model(model_name, X, y, param_grid, n_folds=5)
    all_models_score[model_name] = (score, model)

# Find the best model
best_model_name = max(all_models_score, key=lambda k: all_models_score[k][0])
best_r2_score, best_model = all_models_score[best_model_name]

print(f"\n✅ Best Model: {best_model_name} with R² Score: {best_r2_score:.4f}")
```
