```python
#importing the Libraies
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Reading the Dataset
dataset = pd.read_csv('insurance_pre.csv')
dataset=pd.get_dummies(dataset,drop_first=True)
dataset
```

```python
indep=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
dep=dataset[['charges']]
```

```python
from sklearn.preprocessing import StandardScaler
sc_input = StandardScaler()
sc_output = StandardScaler()
X = sc_input.fit_transform(indep)
y = sc_output.fit_transform(dep)
```

```python
from sklearn.svm import SVR

regressor = SVR(kernel = 'rbf') 
regressor.fit(X, y)
```

```python
from sklearn.metrics import r2_score
y_pred = regressor.predict(X)
r2_value = r2_score(y, y_pred)
print("R2 Score:", r2_value)
```

```python
import pickle
filename="finalized_model_regression.sav"
pickle.dump(regressor,open(filename,'wb'))

sc_input_filename="sc_input.sav"
pickle.dump(sc_input,open(sc_input_filename,'wb'))
sc_output_filename="sc_output.sav"
pickle.dump(sc_output,open(sc_output_filename,'wb'))

```

```python
age_input=45
bmi_input=29
children_input=2
sex_male_input=0
smoker_yes_input=1

loaded_model=pickle.load(open("finalized_model_regression.sav",'rb'))
preprocessed_input = sc_input.transform([[age_input, bmi_input, children_input,sex_male_input,smoker_yes_input]])

pre_prediction=loaded_model.predict(preprocessed_input)

actual_prediction = sc_output.inverse_transform(pre_prediction.reshape(-1, 1))
print('Insurance Charges : Rs.',actual_prediction[0][0].round(2))
```
