```python
import pandas as pd
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
```

```python
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
print(dataset.columns)
dataset
```

```python
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
independent
```

```python
dependent=dataset[["charges"]]
```

```python
X_train,X_test,y_train,y_test=train_test_split(independent, dependent, test_size=0.25,random_state=10)
```

### AdaBoost Algorithm


```python
from sklearn.ensemble import AdaBoostRegressor      # For AdaBoost regression
```

```python
regressor=AdaBoostRegressor(random_state=10, n_estimators=100)
regressor=regressor.fit(X_train,y_train)
```

```python
y_pred=regressor.predict(X_test)
r_score=r2_score(y_test,y_pred)
r_score
```

### XGBoost Algorithm

```python
# !pip install xgboost
from xgboost import XGBRegressor                 # For XGBoost regression
```

```python
regressor=XGBRegressor(random_state=10, n_estimators=100, learning_rate=0.1, max_depth=3)
regressor=regressor.fit(X_train,y_train)
```

```python
y_pred=regressor.predict(X_test)
r_score=r2_score(y_test,y_pred)
r_score
```

### LightGBM Algorithm


```python
# pip install lightgbm
from lightgbm import LGBMRegressor                 # For LightGBM regression
```

```python

regressor=LGBMRegressor(random_state=10, n_estimators=100, learning_rate=0.1, max_depth=3)
regressor=regressor.fit(X_train,y_train)
```

```python
y_pred=regressor.predict(X_test)
r_score=r2_score(y_test,y_pred)
r_score
```
