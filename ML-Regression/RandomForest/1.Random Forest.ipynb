```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
```

```python
dataset=pd.read_csv("50_Startups.csv")
dataset=pd.get_dummies(dataset,drop_first=True)
dataset
```

```python
independent=dataset[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
independent
```

```python
dependent=dataset[["Profit"]]
```

```python
X_train,X_test,y_train,y_test=train_test_split(independent, dependent, test_size=0.30,random_state=0)
```

```python
regressor = RandomForestRegressor(n_estimators = 100, random_state = 10)       
regressor.fit(X_train, y_train)
```

```python
y_pred=regressor.predict(X_test)
```

```python
r_score=r2_score(y_test,y_pred)
r_score
```

```python
import pickle
filename="finalized_model_Forest.sav"
pickle.dump(regressor,open(filename,'wb'))
```

```python
loaded_model=pickle.load(open("finalized_model_Forest.sav",'rb'))
result=loaded_model.predict([[114523.61,153514.11,134050.07,0,1]])
```
