Question 1: Multi-Agent Role Interaction (Reviewer ➝ Editor ➝ Finalizer)

Task: Implement a three-agent pipeline where each agent performs a unique role. Pass the message correctly to avoid format errors.

```python
from autogen import ConversableAgent

config = {
    "config_list":[{"model":"llama3.1:latest",
                    "api_type":"ollama"}]
}

reviewer = ConversableAgent(
    name="Reviewer",
    system_message="Review the text and suggest improvements.",
    llm_config=config
)

editor = ConversableAgent(
    name="Editor",
    system_message="Edit the text based on reviewer comments.",
    llm_config=config
)

finalizer = ConversableAgent(
    name="Finalizer",
    system_message="Produce the final polished version.",
    llm_config=config
)

text = "The project was good but many issue happened."

review = reviewer.generate_reply(
    messages=[{"role":"user","content":text}]
)

edit = editor.generate_reply(
    messages=[{"role":"user","content":review["content"]}]
)

final = finalizer.generate_reply(
    messages=[{"role":"user","content":edit["content"]}]
)

print(final["content"])
```


  Question 4: Email Response Automation using Multiple Agents

   Task:
    Create a 3-agent flow using ConversableAgent and Gemini:
    
●ReaderAgent: Extracts main point from an email

●ResponderAgent: Drafts a reply

●FormatterAgent: Formats it as a proper email


```python
from autogen import ConversableAgent

config = {
    "config_list":[{"model":"llama3.1:latest",
                    "api_type":"ollama"}]
}

reader = ConversableAgent(
    name="ReaderAgent",
    system_message="Extract the main request from the email.",
    llm_config=config
)

responder = ConversableAgent(
    name="ResponderAgent",
    system_message="Write a reply to the email request.",
    llm_config=config
)

formatter = ConversableAgent(
    name="FormatterAgent",
    system_message="Format the reply as a professional email.",
    llm_config=config
)

email = """
Hi Team,
Could you please send the project report by tomorrow?
Thanks
"""

main_point = reader.generate_reply(
    messages=[{"role":"user","content":email}]
)

reply = responder.generate_reply(
    messages=[{"role":"user","content":main_point["content"]}]
)

formatted = formatter.generate_reply(
    messages=[{"role":"user","content":reply["content"]}]
)

print(formatted["content"])
```


  Question 5: Meeting Transcript → Action Points
  
  Task:
  Create a 3-agent pipeline to convert a meeting transcript into actionable tasks.

* ListenerAgent: Extracts main discussion points
* ActionAgent: Converts them into to-do tasks
* DeadlineAgent: Assigns timelines to each task

```python
from autogen import ConversableAgent

config = {
    "config_list":[{"model":"llama3.1:latest",
                    "api_type":"ollama"}]
}

listener = ConversableAgent(
    name="ListenerAgent",
    system_message="Extract key discussion points from meeting transcript.",
    llm_config=config
)

action = ConversableAgent(
    name="ActionAgent",
    system_message="Convert discussion points into tasks.",
    llm_config=config
)

deadline = ConversableAgent(
    name="DeadlineAgent",
    system_message="Assign deadlines to each task.",
    llm_config=config
)

transcript = """
We need to improve marketing strategy.
Prepare quarterly report.
Launch new campaign.
"""

points = listener.generate_reply(
    messages=[{"role":"user","content":transcript}]
)

tasks = action.generate_reply(
    messages=[{"role":"user","content":points["content"]}]
)

final_tasks = deadline.generate_reply(
    messages=[{"role":"user","content":tasks["content"]}]
)

print(final_tasks["content"])
```
