```python
from autogen import ConversableAgent

config_list = [
    {
        "model": "llama3.1:latest",
        "api_type": "ollama"
    }
]

my_config = {
    "config_list": config_list,
    "temperature": 0.2,
    "timeout": 120,
}

question_agent = ConversableAgent(
    name="QuestionAgent",
    system_message="Convert user questions into clear research queries.",
    llm_config=my_config
)

research_agent = ConversableAgent(
    name="ResearchAgent",
    system_message="Provide a detailed technical explanation of the topic.",
    llm_config=my_config
)

explainer_agent = ConversableAgent(
    name="ExplainerAgent",
    system_message="Simplify complex explanations so a beginner can understand them easily.",
    llm_config=my_config
)

# Step 1: Convert question into research query
q = question_agent.generate_reply(
    messages=[{
        "role": "user",
        "content": "Create a research query for: What is Quantum Computing?"
    }]
)

# Step 2: Research explanation
r = research_agent.generate_reply(
    messages=[{
        "role": "user",
        "content": q["content"]
    }]
)

# Step 3: Simplify explanation
e = explainer_agent.generate_reply(
    messages=[{
        "role": "user",
        "content": r["content"]
    }]
)

print("Final simplified answer:\n", e["content"])
```

```python
from autogen import ConversableAgent

config_list = [
    {
        "model": "llama3.1:latest",
        "api_type": "ollama"
    }
]

my_config = {
    "config_list": config_list,
    "temperature": 0.2,
    "timeout": 120,
}

# Agents with defined roles
summarizer = ConversableAgent(
    name="Summarizer",
    system_message="You are an expert text summarizer. Produce concise summaries.",
    llm_config=my_config
)

translator = ConversableAgent(
    name="Translator",
    system_message="You are a professional translator who translates English to Tamil.",
    llm_config=my_config
)

article = """
Global warming refers to the long-term increase in Earth's temperature caused primarily by
human activities such as burning fossil fuels, deforestation, and industrial processes.
These activities release greenhouse gases like carbon dioxide and methane that trap heat
in the atmosphere. The effects include rising sea levels, extreme weather events, and
changes in ecosystems.
"""

# Step 1: Summarization
summary = summarizer.generate_reply(
    messages=[{
        "role": "user",
        "content": f"Summarize the following article:\n{article}"
    }]
)

print("Summary:\n", summary["content"])

# Step 2: Translation
translation = translator.generate_reply(
    messages=[{
        "role": "user",
        "content": f"Translate the following text into Tamil:\n{summary['content']}"
    }]
)

print("Translation:\n", translation["content"])
```

```python
from autogen import ConversableAgent

config_list = [
    {
        "model": "llama3.1:latest",
        "api_type": "ollama"
    }
]

config = {
    "config_list": config_list,
    "temperature": 0.2,
    "timeout": 120
}

reviewer = ConversableAgent(
    name="Reviewer",
    system_message="You are a grammar reviewer. Correct grammar and explain the correction briefly.",
    llm_config=config
)

translator = ConversableAgent(
    name="Translator",
    system_message="You translate English text into Tamil. Return only the translated sentence.",
    llm_config=config
)

assigner = ConversableAgent(
    name="TaskAssigner",
    system_message="You convert text into a simple actionable task.",
    llm_config=config
)

review = reviewer.generate_reply(
    messages=[{"role": "user", "content": "Review this sentence: The cat are on the table."}]
)

translated = translator.generate_reply(
    messages=[{"role": "user", "content": review["content"]}]
)

task = assigner.generate_reply(
    messages=[{"role": "user", "content": translated["content"]}]
)

print("Reviewer:", review["content"])
print("Translator:", translated["content"])
print("TaskAssigner:", task["content"])
```
