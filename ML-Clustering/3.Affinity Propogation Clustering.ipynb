```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
print(dataset.head())
X = dataset.iloc[:, [3, 4]].values
print(X)
```

```python
from sklearn.cluster import AffinityPropagation
afp = AffinityPropagation(random_state=10)
y_afp = afp.fit_predict(X)
```

```python
supervised=pd.DataFrame(dataset)

supervised['Cluster_group']=y_afp

supervised
```

```python
import seaborn as sns
facet = sns.lmplot(data=supervised, x=supervised.columns[3], y=supervised.columns[4],fit_reg=False,legend_out=True)
```

```python
import seaborn as sns
facet = sns.lmplot(data=supervised, x=supervised.columns[3], y=supervised.columns[4], hue=supervised.columns[5], 
                   fit_reg=False, legend=True, legend_out=True)
```
