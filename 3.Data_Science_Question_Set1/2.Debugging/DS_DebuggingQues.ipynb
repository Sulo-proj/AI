```python
import pandas as pd  

data = {'Name': ['Alice', 'Bob', 'Charlie'], 
        'Age': [25, 30, 'Thirty-Five'],  # Thirty-Five is not a number
        'Gender': ['F', 'M', 'M']}  # AxisError: axis 75 is out of bounds for array of dimension 1


df = pd.DataFrame(data)

df['Age'] = pd.to_numeric(df['Age'], errors='coerce') # If 'coerce', then invalid parsing will be set as NaN.

average_age = df['Age'].mean()

print(f"Average Age: {average_age}")
```

```python
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]

# outliers = [i for i in data if zscore(data) > 2]  # ValueError: The truth value of an array with more than one element is ambiguous. Use a.any() or a.all()

zscr=zscore(data)
outliers = [data[i] for i in range(0,len(data)) if zscr[i] > 2]
print('Outliers :',outliers)
```

```python
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]
median = np.median(data)

# data = [median if abs(zscore(x)) > 2 else x for x in data] # AxisError: axis 0 is out of bounds for array of dimension 0

zscr = abs(zscore(data))
data = [median if zscr[x] > 2 else data[x] for x in range(0, len(data)) ]
print(data)
```
