```python
import yfinance as yf
import pandas as pd
```

```python
nifty50_tickers = [
    "HDFCBANK.NS",                  # HDFC Bank
    "ICICIBANK.NS",                 # ICICI Bank
    "RELIANCE.NS",                  # Reliance Industries
    "AXISBANK.NS",                  # Axis Bank
    "BAJFINANCE.NS",                # Bajaj Finance
    "ADANIPORTS.NS",                # Adani Ports & SEZ
    "SBIN.NS",                      # State Bank of India
    "SUNPHARMA.NS",                 # Sun Pharmaceutical Industries
    "BHARTIARTL.NS",                # Bharti Airtel
    "ITC.NS",                       # ITC Limited
    "KOTAKBANK.NS",                 # Kotak Mahindra Bank
    "TATAMOTORS.NS",                # Tata Motors
    "INFY.NS",                      # Infosys
    "TCS.NS",                       # Tata Consultancy Services
    "HCLTECH.NS",                   # HCLTech
    "BEL.NS",                       # Bharat Electronics
    "ONGC.NS",                      # Oil and Natural Gas Corporation
    "NTPC.NS",                      # NTPC Limited
    "EICHERMOT.NS",                 # Eicher Motors
    "LT.NS",                        # Larsen & Toubro
    "COALINDIA.NS",                 # Coal India
    "CIPLA.NS",                     # Cipla
    "GRASIM.NS",                    # Grasim Industries
    "ULTRACEMCO.NS",                # UltraTech Cement
    "TITAN.NS",                     # Titan Company
    "JSWSTEEL.NS",                  # JSW Steel
    "ETERNAL.NS",                   # Eternal Limited
    "POWERGRID.NS",                 # Power Grid Corporation of India
    "TRENT.NS",                     # Trent Limited
    "ADANIENT.NS",                  # Adani Enterprises
    "WIPRO.NS",                     # Wipro
    "MARUTI.NS",                    # Maruti Suzuki
    "HINDALCO.NS",                  # Hindalco Industries
    "HINDUNILVR.NS",                # Hindustan Unilever
    "SBILIFE.NS",                   # “SBI Life Insurance” 
    "TATASTEEL.NS",                 # Tata Steel Ltd
    "HDFCLIFE.NS",                  # “HDFC Life Insurance”
    "DRREDDY.NS",                   # Dr. Reddy’s Laboratories
    "JIOFINANCIAL.NS",              # Jio Financial Services
    "BAJAJFINSV.NS",                # Bajaj Finserv
    "BAJAJ-AUTO.NS",                # Bajaj Auto (or "BAJAJAUTO.NS")
    "ASIANPAINT.NS",                # Asian Paints
    "TATACONSUM.NS",                # Tata Consumer Products
    "TECHM.NS",                     # Tech Mahindra
    "APOLLOHOSP.NS",                # Apollo Hospitals
    "M&M.NS",                       # Mahindra & Mahindra
    "MAXHEALTH.NS",                 # Max Healthcare
    "INDIGO.NS",                    # IndiGo (InterGlobe Aviation)
    "NESTLEIND.NS",                 # Nestlé India
    "SHRIRAMFIN.NS"                 # Shriram Finance Ltd
]


index_tickers = {
    "NIFTY50": "^NSEI",
}
```

```python

start_date = "2014-01-01"
end_date   = "2024-12-31"

all_data = []

for ticker in nifty50_tickers:

    print(f"Downloading {ticker} ...")
    df = yf.download(ticker, start=start_date, end=end_date, multi_level_index=False, auto_adjust=True)


    info_ticker = yf.Ticker(ticker)
    
    if not df.empty:
        df["Stock"] = ticker.replace(".NS", "")
        df['Sector'] = info_ticker.info['sector']
        df['Industry'] = info_ticker.info['industry']
        df['Market Cap'] = info_ticker.info['marketCap']

        df.reset_index(inplace=True)   # Move Date from index to column
        all_data.append(df)


# Combine all stock data
stocks_df = pd.concat(all_data, ignore_index=True)

stocks_df.to_csv("nifty50_stocks.csv", index=False)
stocks_df
```
