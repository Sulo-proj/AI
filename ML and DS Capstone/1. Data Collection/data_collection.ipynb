```python
nifty_tickers=large_tickers + mid_tickers + small_tickers
```

```python
start_date = "2022-01-01"
end_date   = "2024-12-31"

all_data = []

for ticker in nifty_tickers:

    df = yf.download(ticker, start=start_date, end=end_date, multi_level_index=False, auto_adjust=True, progress=False)

    stock = yf.Ticker(ticker)
    info = stock.info
    
    if not df.empty:
        df["Stock"] = ticker.replace(".NS", "")
        df['Sector'] = stock.info.get('sector')
        df['Industry'] = stock.info.get('industry')
        df['Market Cap'] = stock.info.get('marketCap')
        df['Dividend Yield'] = stock.info.get('trailingAnnualDividendYield')
        df['Dividend Rate'] = stock.info.get('trailingAnnualDividendRate')
        df['PE Ratios'] = stock.info.get('trailingPE')
        df['ROE Values'] = stock.info.get('netIncomeToCommon') / (stock.info.get('bookValue') * stock.info.get('sharesOutstanding'))

        if ticker in large_tickers:
            df['Cap Category'] = 'Large Cap'
        elif ticker in mid_tickers:
            df['Cap Category'] = 'Mid Cap'
        elif ticker in small_tickers:
            df['Cap Category'] = 'Small Cap'
        else:
            df['Cap Category'] = None

        df.reset_index(inplace=True)   # Move Date from index to column
        all_data.append(df)

# Combine all stock data
stocks_df = pd.concat(all_data, ignore_index=True)

stocks_df.to_csv("nifty300_stocks.csv", index=False)
stocks_df
```
