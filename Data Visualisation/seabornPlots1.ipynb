```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from data_analysis_utils import Univariate as univ

sns.set_theme(palette='Set1', style='whitegrid', context='notebook')
```

```python
data = pd.read_csv('PrePlacement.csv')
data.drop('sl_no', axis=1, inplace=True)
data
```

### CountPlot

```python
# Univariate
# Categorical / Discrete

sns.countplot(data, x='status', stat='percent')
plt.title('CountPlot for Status in Percentage')
plt.show()
```

```python
# Bivariate
# Categorical + Categorical / Discrete

sns.countplot(data, x='status', hue='workex' , stat='percent')
plt.title('CountPlot for Status based on WorkExp in Percentage')
plt.show()
```

### BarPlot

```python
# Bivariate - Categorical + Numerical

sns.barplot(data,x='workex', y='salary')
plt.title('BarPlot for Salary based on WorkExp')
plt.show()
```

```python
# Multivariate
# Categorical + Categorical + Numerical

sns.barplot(data,x='workex', y='salary', hue='gender', errorbar=None)
plt.title('BarPlot for Salary based on WorkExp and Gender')
plt.show()
```

### BoxPlot

```python
# Univariate
# Numerical

sns.boxplot(data[data['salary']>0], x='salary', width = 0.3)
plt.title('BoxPlot for Salary')
plt.show()
```

```python
# Bivariate
# Categorical + Numerical

sns.boxplot(data[data['salary']>0], x='specialisation', y='salary')
plt.title('BoxPlot of Salary based on Specialisation')
plt.show()
```

```python
# Multivariate
# 2 Categorical + Numerical

sns.boxplot(data[data['salary']>0], x='specialisation', y='salary', hue='degree_t', showmeans=True, meanprops={'marker':'o',
                                                                                                                'markersize':10,
                                                                                                                'markerfacecolor':'white',
                                                                                                                'markeredgecolor':'black'}  )
plt.title('BoxPlot of Salary based on Specialisation and Degree')
plt.legend(bbox_to_anchor=(1,1))
plt.show()
```

```python
# Multivariate
# All Numeric columns

sns.boxplot(data.iloc[:,:-1], showmeans=True, meanprops={'marker':'o',         # Removed the Salary Column
                                                         'markersize':10,
                                                         'markerfacecolor':'white',
                                                         'markeredgecolor':'black'}) 
```

### ViolinPlot

```python
# Univariate
# Numerical

sns.violinplot(data, x='salary')
plt.title('Violin Plot for Salary')
plt.plot()
```

```python
# Bivariate
# Categorical + Numerical

sns.violinplot(data, x='ssc_b', y='salary', inner='quartile')
plt.title('ViolinPlot for Salary based on SSC Board')
plt.show()
```

```python
# Multivariate
# 2 Categorical + 1 Numerical

sns.violinplot(data, x='degree_t', y='salary', hue='hsc_b', split=True)
plt.title('ViolinPlot for Salary based on Degree and HSC Board')
plt.legend(bbox_to_anchor=(1,1))
plt.show()
```

### StripPlot

```python
# Univariate
# Numerical

plt.figure(figsize=(10,5))
sns.stripplot(data, y='salary', jitter=0.4, linewidth=0.5)
plt.title('StripPlot for Salary')
plt.show()
```

```python
# Bivariate
# Categorical + Numerical

plt.figure(figsize=(10,5))
sns.stripplot(data=data, y='salary', hue='specialisation', jitter=0.4, linewidth=0.5)
plt.title('StripPlot for Salary Based on Specialisation')
plt.show()
```

```python
# Multivariate
# 2 Categorical + Numerical

plt.figure(figsize=(10,5))
sns.stripplot(data=data, y='salary', x='specialisation', hue='gender', jitter=0.3, dodge=True, linewidth=0.5)
plt.title('StripPlot for Salary Based on Specialisation and Gender')
plt.show()
```

```python
sns.stripplot(data, x='specialisation', y='salary', jitter=0.3)
sns.violinplot(data, x='specialisation', y='salary', fill=False, inner='quartile')
plt.title('StripPlot and ViolinPlot on Salary based on Specialisation')
plt.show()
```

```python
sns.stripplot(data, x='specialisation', y='salary', jitter=0.3)
sns.boxplot(data, x='specialisation', y='salary', fill=False)
plt.title('StripPlot and BoxPlot on Salary based on Specialisation')
plt.show()
```

### SwarmPlot

```python
# Univariate
# Numerical

sns.swarmplot(data, y='salary', size=4)
plt.title('SwarmPlot for Salary')
plt.show()
```

```python
# Bivariate
# Categorical + Numerical

sns.swarmplot(data, x='specialisation', y='salary', size=3)
plt.title('SwarmPlot for Salary based on Specialisation')
plt.show()
```

```python
# Multivariate
# 2 Categorical + Numerical

plt.figure(figsize=(10,5))
sns.swarmplot(data=data, x='specialisation', y='salary', hue='degree_t', dodge=True, size=4)
plt.title('SwarmPlot for Salary based on Specialisation and Degree')
plt.legend(bbox_to_anchor=(1, 1), loc='upper left')
plt.show()
```

```python


plt.figure(figsize=(10,5))
sns.swarmplot(data=data, x='specialisation', y='salary', hue='degree_t', dodge=True, size=4)
sns.violinplot(data, x='specialisation', y='salary', hue='degree_t', fill=False, inner='quartile', bw_adjust=0.7)
plt.title('ViolinPlot and SwarmPlot for Salary based on Specialisation and Degree')
plt.legend(bbox_to_anchor=(1, 1), loc='upper left')
plt.show()
```

### FactorPlot / CatPlot

```python
# Multivariate
# Categorical + Numerical

sns.catplot(data=data, x='status', y='mba_p', col='gender', row='specialisation', hue='workex', kind='violin')
plt.show()
```

### Histogram

```python
# Univariate
# Numeric - Continuous

sns.histplot(data, x='salary', binwidth=50000, kde=True)
plt.title('Histogram of Salary')
plt.show()
```

```python
# Bivariate
# Numeric + Categoric


sns.histplot(data, x='salary', hue='workex', bins=25, kde=True, multiple='dodge', stat='probability')
plt.title('Histogram of Salary')
plt.show()
```

```python
# Bivariate
# Numeric + Numeric


sns.histplot(data, x='salary', y='etest_p', bins=25)
plt.title('Histogram of Salary Vs ETest Percent')
plt.show()
```

### KDE Plot

```python
# Univariate
# Numeric

sns.kdeplot(data, x='etest_p')
plt.title('Density Plot for ETest Percent')
plt.show()
```

```python
# Bivariate
# Numeric + Categorical

sns.kdeplot(data, x='salary', hue='specialisation')
plt.title('Density Plot for Salary based on Specialisation')
plt.show()
```

```python
# Bivariate
# 2 Numeric

sns.kdeplot(data, x='salary', y='etest_p')
plt.title('Density Plot for Salary Vs ETest Percent')
plt.show()
```

```python
# Multivariate
# All Numeric

sns.kdeplot(data.iloc[:,:-1])
plt.title('Density Plot for All Academic Scores')
plt.show()
```

### RugPlot

```python
# Univariate
# Numeric

plt.figure(figsize=(12,5))
sns.rugplot(data, x='salary')
plt.title('RugPlot for Salary')
plt.show()
```

```python
# Bivariate
# Numeric

plt.figure(figsize=(10,10))
sns.rugplot(data, x='etest_p', y='mba_p', height=0.1)
plt.title('RugPlot for Etest Vs MBA Percent')
plt.show()
```

```python
# Multivariate
# 2 Numeric + Categoric

plt.figure(figsize=(10,10))
sns.rugplot(data, x='etest_p', y='mba_p', hue='gender', height=0.1)
sns.scatterplot(data, x='etest_p', y='mba_p', hue='gender')
plt.title('RugPlot for Etest Vs MBA Percent')
plt.show()
```
