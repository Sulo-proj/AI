## 4. Cleaning Dataset

### 1. Remove Unnecessary Columns

```python

# removing unwanted columns
raw_data=raw_data.drop(columns=['Zomato URL', 'Address'], axis=1)
raw_data
```

### 2. Rating and Votes Column

We will add Delivery and Dining Rating into one Rating column - to rate out of 10

Also, Delivery and Dining Rating Count into one Votes column

```python
raw_data['Rating'] = raw_data['Delivery Rating']+ raw_data['Dining Rating']
raw_data['Votes'] = raw_data['Delivery Rating Count']+ raw_data['Dining Rating Count']
raw_data.drop(columns=['Delivery Rating', 'Dining Rating', 'Delivery Rating Count', 'Dining Rating Count'], inplace=True)
raw_data
```

```python
raw_data['Votes'] = pd.to_numeric(raw_data['Votes'], errors='coerce')
```

### 3. Removing the NaN records 

```python
# remove null columns
raw_data.isnull().sum()
```

```python
raw_data.dropna(inplace=True)
raw_data
```

```python
raw_data.isnull().sum()
```

### 4. Working on 'Location' Column

```python
raw_data['Location'].value_counts()

```

```python
raw_data['Location'] = raw_data['Location'].str.replace('[A-Za-z ]+, ', '', regex=True)
raw_data['Location'] = raw_data['Location'].str.replace('[A-Za-z ]+,', '', regex=True)
loc_list = raw_data['Location'].value_counts(ascending=True)
loc_list.to_csv('Location1.csv')
loc_list
```

```python
region_map = {
    r'A Radha Hotel': 'Manapakkam',
    r'V7 Hotel': 'Porur',
    r'Abhiramapuram': 'Mylapore',
    r"Taj Fisherman's Cove Resort &|Kanchipuram District": 'Kanchipuram',
    r"The King'Egmore": 'Egmore',
    r'Somerset Greenways|MRC Nagar': 'RA Puram',
    r'Grande Bay Resort|Mamallapuram': 'Mahabalipuram ',
    r'OMR Food Street,Thuraipakkam': 'Thuraipakkam',
    r'Grand by GRT Hotels|T.Nagar': 'T. Nagar',
    r'Novotel Chennai Sipcot|Novotel Chennai Sipcot': 'Sipcot',
    r'East...|East Coast Road (ECR)': 'ECR',
    r'Citadines|Old Mahabalipuram Road (OMR)|Dash@Old Mahabalipuram Road (OMR)|Gokulam Park Sabari -Old Mahabalipuram Road|Holiday Inn Chennai OMR IT Expressway|Turyaa Chennai|Gokulam Park Sabari -Old Mahabalipuram Road': 'OMR',
    r'RK Salai (Cathedral Road)': 'Gopalapuram',
    r'The Slate': 'Nungambakkam'
}

# one-shot replacement
raw_data.replace({'Location':region_map}, regex=True, inplace=True)
loc_list = raw_data['Location'].value_counts(ascending=True)
loc_list.to_csv('Location1.csv')
loc_list
```

### 6. Working on Cuisine Column - Create Cuisine Type with first Cuisine on the lsit

```python
# Extract the first cuisine from the string
raw_data['Cuisine Type'] = raw_data['Cuisine'].str.extract(r"\['?\s*([^',\]]+)")

raw_data
```

```python
raw_data['Cuisine Type'].value_counts()
```

### 7. Copying the cleaned data into a new DataFrame

```python
zomato_data = raw_data.reset_index(drop=True).copy()
zomato_data
```

## 5. Performing Exploratory Data Analysis

### Q1. Top 10 restaurants with most branches in Chennai

```python
top_branches = zomato_data.groupby(['Name of Restaurant'])['Location'].count().sort_values(ascending=False).reset_index()
top_branches = top_branches.rename(columns={'Location':'Count'}).head(10)
top_branches
```

### Q10. Find the list of Affordable Restaurants?

The criteria for Affordable Restaurants would be:-
1) Low Price
2) High Rated

First step will be to find the restaurants with average cost 1/4th the average cost of most expensive restaurant in our dataframe.

Then find restaurants with more than 9 Rating from this list

```python
max_price = zomato_data['Price for 2'].max()
one_fourth=int(max_price/4)
one_fourth
```

```python
affordable_df = zomato_data[zomato_data['Price for 2'] <= one_fourth]
affordable_df=affordable_df.sort_values(by='Price for 2').reset_index(drop=True)
affordable_df
```

```python
rating_high_affortable_df = affordable_df[affordable_df['Rating'] >= 9].reset_index(drop=True)
rating_high_affortable_df
```

### Q11. Find the list of most Reliable Restaurants?

The criteria for most Reliable Restaurants would be:-
1) Low Price ( less than 1/4th highest Price )
2) High Rated ( higher than 9 )
3) Large No. of Votes ( more than mean Votes )

```python
# Low Price (Less than 1/4th Highest Price)
affordable_price = int(zomato_data['Price for 2'].max()/4)
reliable_rest_df = zomato_data[zomato_data['Price for 2'] <= affordable_price]
reliable_rest_df
```

```python
# Highest Rated
reliable_rest_df = reliable_rest_df[reliable_rest_df['Rating'] >= 9]
reliable_rest_df
```

```python
# Large no of votes
mean_votes = int(zomato_data['Votes'].mean())
reliable_rest_df = reliable_rest_df[reliable_rest_df['Votes'] >= mean_votes].sort_values(by='Votes', ascending=False).reset_index(drop=True)
reliable_rest_df
```
