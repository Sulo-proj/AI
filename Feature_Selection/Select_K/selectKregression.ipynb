## Import Modules and Dataset

```python
import pandas as pd
import data_analysis_utils as utils
import selectKregression as k_regr

inp_dataset = pd.read_csv('kidney_disease.csv')
inp_dataset = pd.read_csv('kidney_disease.csv').drop(columns='id')

inp_dataset
```

## Analysing Raw data and Preprocess

```python
print('Missing Value Percentages:', inp_dataset.isnull().mean()*100)

quan, qual = utils.Preprocessing.quanQual(inp_dataset)
inp_dataset = utils.Preprocessing.quanVariables(inp_dataset, qual)
```

```python
inp_dataset[['pcv', 'rc', 'wc']]=inp_dataset[['pcv', 'rc', 'wc']].astype(float)
quan, qual = utils.Preprocessing.quanQual(inp_dataset)
```

```python
inp_dataset = utils.Preprocessing.simple_missing(inp_dataset, quan, qual)
inp_dataset = utils.Preprocessing.model_missing(inp_dataset, quan)
```

## Input, Target variables split

```python
data = pd.get_dummies(inp_dataset, drop_first=True)
indep_X = data.iloc[:, :-1]
dep_Y = data.iloc[:, -1]
```

## Feature Selection

```python
fullresult = pd.DataFrame()

for n in range(3, 6):
    df_results, selected_features = k_regr.run_regressors(indep_X, dep_Y, k_no=n)
    fullresult = pd.concat([fullresult, df_results], ignore_index=True)

# Format float columns to 4 decimal places for display
float_cols = [col for col in fullresult.columns if col.endswith('_R2')]
fullresult[float_cols] = fullresult[float_cols].round(4)
fullresult
```

```python
max_Accuracy = fullresult.iloc[:,2:].values.max()

max_pos = fullresult.iloc[:, 2:].stack().idxmax()  # (row_index, column_name)

row_index = max_pos[0]
Model_Name = max_pos[1]

k_No = fullresult.iloc[row_index, 0]
select_features = fullresult.iloc[row_index, 1]

print(f'Max Accuracy is: {max_Accuracy:.4f}% for {Model_Name} Model \nWith K = {k_No} and selected Features: {select_features}')

```
