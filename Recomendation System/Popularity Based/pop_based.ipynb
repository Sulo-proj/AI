```python
# importing libraries
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

#Read the movie Dataset
i_cols = ['movie_id', 'movie_title' ,'release_date','video_release date', 'IMDb_URL', 'unknown', 'Action', 'Adventure',
    'Animation', 'Children\'s', 'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy',
    'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']
movies = pd.read_csv('ml-100k/u.item', sep='|', names=i_cols,encoding='latin-1')

# reading ratings file:
r_cols = ['user_id', 'movie_id', 'rating', 'unix_timestamp']
ratings = pd.read_csv('ml-100k/u.data', sep='\t', names=r_cols,encoding='latin-1')

# Specify genre columns
genre_columns =['Action', 'Adventure', 'Animation', 'Children\'s', 'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy',
    'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']
movies.head(3)
```

```python
# We give 1 for every rating from user - no of times any movie is watched and rated makes it popular.
ratings['popularity']=1
ratings.drop(columns=['rating', 'user_id', 'unix_timestamp'], axis=1, inplace=True)
popularity_df = ratings.groupby('movie_id').sum().reset_index()
popularity_df.set_index('movie_id', inplace=True)
mov_ind = movies.set_index('movie_id')

popularity_df
```

```python
pop_mov_df = popularity_df.join(mov_ind, on='movie_id', how='inner')
pop_sorted = pop_mov_df.sort_values(by='popularity', ascending=False)
pop_sorted
```

```python

for genre in genre_columns:
    print(f'\nThe Top 5 Popular movies in {genre} genre :\n')
    top5_genre = pop_sorted[pop_sorted[genre]>0].head(5)['movie_title'].values
    for top5 in top5_genre:
        print(top5)
```
