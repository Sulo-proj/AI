## Language Tone Converter Agent Chain

```python
from autogen import ConversableAgent

config = {
    "config_list": [{
        "model": "llama3.1:latest",
        "api_type": "ollama"
    }],
    'temperature': 0.4,
    'timeout':120
}

toneAgent = ConversableAgent(
    name="ToneAgent",
    system_message='You are tone changer. Change the tone to more polite',
    llm_config=config
    )
message = "can u send me the file soon?"

tone_adjust = toneAgent.generate_reply(messages=[{"role": "user", "content": f'Change the tone of this \n{message}'}])
print(tone_adjust["content"])
```

## Summary Generator for Meeting Notes

## Improper Agent Chaining – Summary and Sentiment

```python
from autogen import ConversableAgent

config = {
    "config_list": [{
        "model": "llama3.1:latest",
        "api_type": "ollama"
    }],
    'temperature':0.8,
    'timeout':100
}

summarizer = ConversableAgent(
    name="Summarizer", 
    system_message='You are Summarizer Assistant. You summarize the given message.',
    llm_config=config)
sentimenter = ConversableAgent(
    name="SentimentAnalyzer", 
    system_message='You are Sentiment Analyzer bot. You will Analyze the Sentiment of the given message and give reason.',
    llm_config=config)

review_text = "I ordered a laptop last week and it arrived with a broken screen. Extremely disappointed!"

summary = summarizer.generate_reply(messages=[{"role": "user", "content": review_text}])
sentiment = sentimenter.generate_reply(messages=[{"role": "user", "content": summary['content']}])  

print("Summary:\n", summary["content"])
print("Sentiment:\n", sentiment["content"])
```

## Task Assignment Confusion

```python
from autogen import ConversableAgent

config = {
    "config_list": [{
        "model": "llama3.1:latest",
        "api_type": "ollama"
    }],
    'temperature':0.8,
    'timeout':100}

worker = ConversableAgent(
    name="TaskAgent", 
    system_message='You are Task Agent. You will Assign Tasks based on the messages.',
    llm_config=config)
planner = ConversableAgent(
    name="PlanAgent", 
    system_message='You are Planning Agent. You will Plan steps based on messages.',
    llm_config=config)  

task_plan = planner.generate_reply(messages=[{"role": "user", "content": "Plan steps to clean and organize my digital files"}])
execution = worker.generate_reply(messages=[{"role": "user", "content": task_plan["content"]}])

print("Plan:", task_plan["content"])
print("Execution:", execution["content"])
```

## Reviewer–Approver (Blog Publishing)

```python
from autogen import ConversableAgent

config = {
    "config_list": [{
        "model": "llama3.1:latest",
        "api_type": "ollama"
    }],
    'temperature':0.8,
    'timeout':100
}

reviewer = ConversableAgent(
    name="ReviewerAgent",
    system_message='You are Blog Reviewer Assistant. You review the blogs and suggests improvements.',
    llm_config=config)
approver = ConversableAgent(
    name="ApproverAgent",
    system_message='You are Approver Agent. You Approve the blog if it is ready to publish. Specify with "Yes" or "No" with a short simple reason', 
    llm_config=config)  

blog_draft = "AI is good. It helps. Many use it."

review_feedback = reviewer.generate_reply(messages=[{"role": "user", "content": f"Review this blog and suggest improvements: {blog_draft}"}])
approval = approver.generate_reply(messages=[{"role": "user", "content": review_feedback["content"]}])

print("Feedback:", review_feedback["content"])
print("Approval:", approval["content"])
```
