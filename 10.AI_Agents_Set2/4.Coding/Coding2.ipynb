Question 1: Sequential Task Pipeline

Task: Implement a 4-step document workflow using Gemini-powered ConversableAgents. The steps are:

* DraftWriter → Writes a short article draft on a given topic.
* EditorAgent → Refines the grammar, structure, and tone of the draft.
* PlagiarismChecker → Checks if the content is original or copied.
* PublisherAgent → Finalizes the article and prepares it for publishing.

Each agent should take the previous agent's output and continue the task without losing context.  Print each step’s output clearly. 

Use generate_reply() for all interactions.

```python
from autogen import ConversableAgent

config = {
    "config_list":[{"model":"llama3.1:latest",
                    "api_type":"ollama"}]
}

draft_writer = ConversableAgent(
    name="DraftWriter",
    system_message="Write a short article draft based on the given topic.",
    llm_config=config
)

editor = ConversableAgent(
    name="EditorAgent",
    system_message="Improve grammar, structure, and tone of the article.",
    llm_config=config
)

plagiarism = ConversableAgent(
    name="PlagiarismChecker",
    system_message="Check whether the article appears original or copied.",
    llm_config=config
)

publisher = ConversableAgent(
    name="PublisherAgent",
    system_message="Prepare the article in final publish-ready format.",
    llm_config=config
)

topic = "The importance of AI in healthcare"

draft = draft_writer.generate_reply(
    messages=[{"role":"user","content":topic}]
)

edited = editor.generate_reply(
    messages=[{"role":"user","content":draft["content"]}]
)

checked = plagiarism.generate_reply(
    messages=[{"role":"user","content":edited["content"]}]
)

final = publisher.generate_reply(
    messages=[{"role":"user","content":edited["content"]}]
)

print("\n--- Draft ---\n", draft["content"])
print("\n--- Edited Article ---\n", edited["content"])
print("\n--- Plagiarism Check ---\n", checked["content"])
print("\n--- Final Published Article ---\n", final["content"])
```

Question 2. Task Decomposer (Planner + Workers)

Use a PlannerAgent to break down the goal: "Create a business plan for a startup in mental health tech". Then use MarketAgent, TechAgent, and FinancialAgent to address each part.

Question 4. Multi-Language Translator Agents

Question: Input a paragraph in English. EnglishToSpanish, SpanishToFrench, and FrenchToGerman agents must translate it in chain. Output should be the final German translation.

```python
from autogen import ConversableAgent

config = {
    "config_list":[{"model":"llama3.1:latest",
                    "api_type":"ollama"}]
}

eng_sp = ConversableAgent(
    name="EnglishToSpanish",
    system_message="Translate English to Spanish.",
    llm_config=config
)

sp_fr = ConversableAgent(
    name="SpanishToFrench",
    system_message="Translate Spanish to French.",
    llm_config=config
)

fr_de = ConversableAgent(
    name="FrenchToGerman",
    system_message="Translate French to German.",
    llm_config=config
)

text = "Artificial intelligence is transforming modern healthcare."

spanish = eng_sp.generate_reply(
    messages=[{"role":"user","content":text}]
)

french = sp_fr.generate_reply(
    messages=[{"role":"user","content":spanish["content"]}]
)

german = fr_de.generate_reply(
    messages=[{"role":"user","content":french["content"]}]
)

print("\n--- Final German Translation ---\n", german["content"])
```

Question 5: Fact Checker Agent Flow (Truth vs. Fake Detection)

Task: Create a pipeline to verify a social media claim using 3 agents:
* ClaimExtractorAgent: Extracts main claim from user input
* FactCheckerAgent: Determines if it's real or fake
* ReporterAgent: Outputs a news-style response

```python
from autogen import ConversableAgent

config = {
    "config_list":[{"model":"llama3.1:latest",
                    "api_type":"ollama"}]
}

extractor = ConversableAgent(
    name="ClaimExtractorAgent",
    system_message="Extract the main factual claim from the input.",
    llm_config=config
)

checker = ConversableAgent(
    name="FactCheckerAgent",
    system_message="Determine if the claim is likely true or fake.",
    llm_config=config
)

reporter = ConversableAgent(
    name="ReporterAgent",
    system_message="Write a short news-style report about the claim verification.",
    llm_config=config
)

post = "A social media post claims that drinking hot water cures all viral infections."

claim = extractor.generate_reply(
    messages=[{"role":"user","content":post}]
)

fact = checker.generate_reply(
    messages=[{"role":"user","content":claim["content"]}]
)

report = reporter.generate_reply(
    messages=[{"role":"user","content":fact["content"]}]
)


print("\n--- Extracted Claim ---\n", claim["content"])
print("\n--- Fact Check Result ---\n", fact["content"])
print("\n--- Final Report ---\n", report["content"])
```
